### ตรวจสอบจำนวน Reviewer ของแต่ละโมเดล

```python
import pandas as pd
from IPython.display import display

file_path = (
    "/Users/chutinun/Data guardrail/"
    "Pre-data/guardrail_combined_dataset.csv"
)

# อ่านเฉพาะคอลัมน์ที่ต้องใช้จากข้อมูลทั้งหมด
reviewer_data = pd.read_csv(
    file_path,
    usecols=["model", "reviewed_by"],
    encoding="utf-8-sig"
)

# เปลี่ยนข้อความว่างให้เป็นค่าว่างมาตรฐาน
reviewer_data["reviewed_by"] = (
    reviewer_data["reviewed_by"]
    .replace(r"^\s*$", pd.NA, regex=True)
)

# เลือกเฉพาะรายการที่มี Reviewer
reviewed = reviewer_data.dropna(subset=["reviewed_by"])

# สรุปจำนวน Reviewer และจำนวนแถวที่ตรวจในแต่ละโมเดล
reviewer_summary = (
    reviewed.groupby("model")
    .agg(
        จำนวน_reviewer=("reviewed_by", "nunique"),
        จำนวนรายการที่ตรวจ=("reviewed_by", "size"),
    )
    .reset_index()
    .sort_values("model")
)

display(reviewer_summary)
```

```python
reviewer_matrix = pd.crosstab(
    reviewed["model"],
    reviewed["reviewed_by"],
    margins=True,
    margins_name="รวม",
)

display(reviewer_matrix)
```

```python
unreviewed_summary = pd.crosstab(
    reviewer_data["model"],
    reviewer_data["reviewed_by"].notna()
    .map({True: "ตรวจแล้ว", False: "ยังไม่มี Reviewer"}),
    margins=True,
    margins_name="รวม"
)

display(unreviewed_summary)
```

## ตรวจสอบจำนวนข้อมูลของแต่ละโมเดล แยกตามประเภท `flag_type`

```python
import pandas as pd
from IPython.display import display

file_path = (
    "/Users/chutinun/Data guardrail/"
    "Pre-data/guardrail_combined_dataset.csv"
)

# อ่านข้อมูลทั้งหมดเฉพาะคอลัมน์ที่ต้องใช้
model_flag = pd.read_csv(
    file_path,
    usecols=["model", "flag_type"],
    encoding="utf-8-sig"
)

# ตารางไขว้: แถวเป็นโมเดล คอลัมน์เป็น flag_type
model_flag_count = pd.crosstab(
    index=model_flag["model"],
    columns=model_flag["flag_type"],
    margins=True,
    margins_name="รวม",
)

display(model_flag_count)
```

### ตรวจสอบจำนวน final_verdict แยกตามโมเดลและประเภท

#### 1. สัดส่วน final_verdict เป็นเปอร์เซ็นต์ โดย flag_type

```python
check_data = pd.read_csv(
    file_path,
    usecols=["model", "flag_type", "final_verdict"],
    encoding="utf-8-sig",
)

check_data["final_verdict"] = (
    check_data["final_verdict"]
    .fillna("<blank>")
    .replace(r"^\s*$", "<blank>", regex=True)
)

verdict_summary = pd.crosstab(
    index=[
        check_data["model"],
        check_data["flag_type"],
    ],
    columns=check_data["final_verdict"],
    margins=True,
    margins_name="รวม",
)

verdict_summary["accept (%)"] = (
    verdict_summary["accept"]
    / (verdict_summary["accept"] + verdict_summary["fail"])
    * 100
).round(2)

verdict_summary["fail (%)"] = (
    100 - verdict_summary["accept (%)"]
).round(2)

display(verdict_summary)
```

#### 2. ความครอบคลุมของ Reviewer

```python
import pandas as pd
from IPython.display import display

file_path = (
    "/Users/chutinun/Data guardrail/"
    "Pre-data/guardrail_combined_dataset.csv"
)

review_data = pd.read_csv(
    file_path,
    usecols=["model", "flag_type", "reviewed_by"],
    encoding="utf-8-sig",
)

# ทำให้ช่องว่างและข้อความว่างเป็น NA
review_data["reviewed_by"] = (
    review_data["reviewed_by"]
    .astype("string")
    .str.strip()
    .replace("", pd.NA)
)

grouped = review_data.groupby(
    ["model", "flag_type"],
    dropna=False,
)

# สร้างตารางทีละคอลัมน์ เพื่อรองรับ pandas หลายเวอร์ชัน
review_coverage = grouped.size().rename("total").to_frame()

review_coverage["reviewed"] = (
    grouped["reviewed_by"].count()
)

review_coverage["reviewer_count"] = (
    grouped["reviewed_by"].nunique()
)

review_coverage["unreviewed"] = (
    review_coverage["total"]
    - review_coverage["reviewed"]
)

review_coverage["coverage_percent"] = (
    review_coverage["reviewed"]
    / review_coverage["total"]
    * 100
).round(2)

review_coverage = (
    review_coverage
    .reset_index()
    .rename(
        columns={
            "model": "โมเดล",
            "flag_type": "ประเภท Flag",
            "total": "จำนวนทั้งหมด",
            "reviewed": "ตรวจแล้ว",
            "reviewer_count": "จำนวน Reviewer",
            "unreviewed": "ยังไม่ตรวจ",
            "coverage_percent": "เปอร์เซ็นต์ที่ตรวจ",
        }
    )
)

display(review_coverage)
```

### จำนวน Final Verdict ของแต่ละโมเดล

```python
verdict_data = pd.read_csv(
    file_path,
    usecols=["model", "final_verdict"],
    encoding="utf-8-sig",
)

verdict_data["final_verdict"] = (
    verdict_data["final_verdict"].fillna("<blank>")
)

verdict_summary = pd.crosstab(
    verdict_data["model"],
    verdict_data["final_verdict"],
    margins=True,
    margins_name="รวม",
)

verdict_summary["accept (%)"] = (
    verdict_summary["accept"]
    / (verdict_summary["accept"] + verdict_summary["fail"])
    * 100
).round(2)

verdict_summary["fail (%)"] = (
    100 - verdict_summary["accept (%)"]
).round(2)

display(verdict_summary)
```
